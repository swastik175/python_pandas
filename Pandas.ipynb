```python
bios_new = bios.copy()
bios_new['height_category'] = bios_new['height_cm'].apply(lambda x: 'Short' if x < 165 else ('Average' if x < 185 else 'Tall'))
```

```python
bios_new
```

```python
def athlete_category(row):
  if row['height_cm'] < 175 and row['weight_kg'] < 70:
    return 'Lightweight'
  elif row['height_cm'] < 185 or row['weight_kg'] >= 70:
    return 'Middleweight'
  else:
    return 'Heavyweight'
bios_new['Category'] = bios_new.apply(athlete_category, axis=1)
bios_new
```

Merging and Concatenating

```python
nocs = pd.read_csv('https://raw.githubusercontent.com/KeithGalli/complete-pandas-tutorial/refs/heads/master/data/noc_regions.csv')
nocs.head()
```

```python
bios_new = pd.merge(bios, nocs, left_on='born_country', right_on='NOC', how='left')
bios_new.rename(columns={'region':'born_country_full'}, inplace=True)
bios_new.head()
```

```python
usa = bios[bios['born_country'] == 'USA'].copy()
gbr = bios[bios['born_country']== 'GBR'].copy()

new_df = pd.concat([usa, gbr])
new_df
```

```python
import pandas as pd
import numpy as np
coffee = pd.read_csv('https://raw.githubusercontent.com/KeithGalli/complete-pandas-tutorial/refs/heads/master/warmup-data/coffee.csv')
coffee.loc[[0,1] , 'Units Sold'] = np.nan
coffee['Units Sold'] = coffee['Units Sold'].interpolate()
coffee['Units Sold'].fillna(10, inplace=True)
coffee.groupby(['Coffee Type'])['Units Sold'].sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count Olympians per country
country_counts = bios['born_country'].value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=country_counts.index, y=country_counts.values, palette='viridis')
plt.yscale('log')
plt.title('Top 10 Countries by Number of Olympians (Log Scale)')
plt.xlabel('Country')
plt.ylabel('Number of Olympians (Log Scale)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
