```python
bios_new = bios.copy()
bios_new['height_category'] = bios_new['height_cm'].apply(lambda x: 'Short' if x < 165 else ('Average' if x < 185 else 'Tall'))
```

```python
bios_new
```

```python
def athlete_category(row):
  if row['height_cm'] < 175 and row['weight_kg'] < 70:
    return 'Lightweight'
  elif row['height_cm'] < 185 or row['weight_kg'] >= 70:
    return 'Middleweight'
  else:
    return 'Heavyweight'
bios_new['Category'] = bios_new.apply(athlete_category, axis=1)
bios_new
```
